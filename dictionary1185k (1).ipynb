```python
#write a program to iterate over a dictionary using for loop and print the keys alone,values alone and both keys and values.

student = {
    "1": "a",
    "2": "b",
    "3": "c",
    "4": "d",
}

print("Keys:")
for key in student:
    print(key)

print("\nValues:")
for value in student.values():
    print(value)

print("\nKeys and Values:")
for key, value in student.items():
    print(key, ":", value)
```

```python
#write a program to prepare a dictionary where the keys are numbers between 1 and 15(both included) and the values are sq of the keys

dict5 = {}
for i in range (1,16):
    dict5[i] = i**2
print(dict5)
```

```python
#write a program to print the mean and sum of all the values in a dictionary,considering the values will of int type
diction = {1:10,2:20,3:30}
total=0
for value in diction.values():
    total = total + value
print("Sum of values:",total)
mean_value = sum(diction.values()) / len(diction)
print("Mean:",mean_value)
```

```python
#display the student with highest marks
mar={"raj":90,"ema":88,"veer":22,"pari":93,"janhavi":94}
highest=max(mar,key=mar.get)
print("Student with the highest marks:",highest)
print("marks scored:",mar[highest])
```

```python
#sort dictionary by its values


stud = {"raj": 90, "ema": 88, "veer": 22, "pari": 93, "janhavi": 94}

# Adding [1] ensures Python looks at the dictionary values (scores)
S = dict(sorted(stud.items(), key=lambda item: item[1]))

print(S)
```

```python
studY={"raj":90,"ema":88,"veer":22,"pari":93,"janhavi":94}
Y=dict(sorted(studY.items(),reverse=True,key=lambda item:item[1]))
print(Y)
```
