```python
#prime number using user efined function
def prime():
    c = 0
    num = int(input("entter a number"))
    for n in range(1,num):
        if num % n == 0:
            c = c +  1
            continue
    if c == 2:
        print("prime")
    else:
        print("not prime")
       

prime()
```

```python
#WAP to add all elements in array() or list[]
def sum(n):
    tot=0
    for x in n:
        tot+=x
    return tot
print(sum((2,4,5,9,7,5)))
```

```python
#wap to reverse a string
def rev(str1):
    rstr=" "
    index = len(str1)
    while index>0:
        rstr+=str1[index - 1] #rstr index increments
        index-=1 #str1 index decrements and is added to rstr1 in next iteration
    return rstr
print(rev("golden"))
```

```python
#WAP to multiply lsit elements
def mut(w):
    
    prod=1
    for y in w:
        prod=prod*y
    return prod
print(mut((3,5,6,7,8,9)))
```

```python
#WPF to find fact
def fact(d):
    f=1
    for i in range(1,d+1):
        f=f*i
    return f
print(fact(4))
```

```python
#to print only even numbers in a list
def even(s):
    for i in s:
        if i % 2 == 0:
            print(i)
    
even((2,3,4,5,6))
    
```

```python
def is_palindrome(str1):
    rstr1=" "
    index = len(str1)
    while index>0:
        rstr1 += str1[index-1]
        index = index-1
    if rstr1==str1:
        print("palidrome")
    else:
        print("not palindrome")


is_palindrome("STAATS")
        
```
