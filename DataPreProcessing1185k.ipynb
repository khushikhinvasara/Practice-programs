```python
#part of assignment 6
#data preprocessing and cleaning
#dropna called as dropNA drops the entire rw evn if it hs only one null/none value
import pandas  as pd
data={
    'a':[1,3,4,None,5],
    'b':[1,2,None,None,6]
}
df=pd.DataFrame(data)
print(df)
print("\n\n")
dfc=df.dropna()
print(dfc)
```

```python
#unlike dropna,fillna doesnt clean or remove the whole row,it replaces the None values with 0 or wtv is declared
data={
    'a':[1,3,4,None,5],
    'b':[1,2,None,None,6],
    'c':[None,2,4,5,None]
}
df=pd.DataFrame(data)
print(df)
print("\n\n")
df.fillna(0,inplace=True)
print(df)
```

```python
data={
    'a':[1,3,4,None,5],
    'b':[1,2,None,None,6],
    'c':[None,2,4,5,None]
}
df=pd.DataFrame(data)
print(df)
print("\n\n")
df.fillna(df.mean(),inplace=True)#df.mean()is an aggregate fn that here replaced None values with mean of the columns
print(df)
```

```python
data={
    'a':[1,3,4,None,5],
    'b':[1,2,None,None,6],
    'c':[None,2,4,5,None]
}
df=pd.DataFrame(data)
print(df)
print("\n\n")
df.rename(columns={'a':'age','b':'name','c':'salary'})
inplace=True
print(df)
```
