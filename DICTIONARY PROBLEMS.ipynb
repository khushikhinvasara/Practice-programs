```python
#write a program to iterate over a dictionary using for loop and print the keys alone,values alone and both keys and values.

student = {
    "1": "a",
    "2": "b",
    "3": "c",
    "4": "d",
}

print("Keys:")
for key in student:
    print(key)

print("\nValues:")
for value in student.values():
    print(value)

print("\nKeys and Values:")
for key, value in student.items():
    print(key, ":", value)
```

```python
#write a program to prepare a dictionary where the keys are numbers between 1 and 15(both included) and the values are sq of the keys

dict = {}
for i in range (1,16):
    dict[i] = i**2
print(dict)
```

```python
#write a program to sum all the values in a dictionary,considering the values will of int type
dict = {1:10,2:20,3:30}
total=0
for value in dict.values():
    total = total + value
print("Sum of values:",total)
```
