```python
#wap to iterate over dictionary using for loop and print keys alone,values alone and both tgt
a={101:"s",102:"e",103:"w"}
print("keys:")
for key in a.keys():
    print(key)
print("\nvalues")
for value in a.values():
    print(value)
print("\nkeys n values")
for key,value in a.items():
    print(key,value)
```

```python
#sum of values in th dictionary
f={"a":10,"b":20,"c":40,"d":22}
total=0
for v in f.values():
    total+=v
print(total)
```
